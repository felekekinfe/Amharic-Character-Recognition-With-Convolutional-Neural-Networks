```python
from PIL import Image, ImageDraw, ImageFont
import os
import random

# Define the Amharic alphabet (33 consonants, 7 vowel forms each)
amharic_chars = []
base_consonants = [
    0x1200,  # ሀ
    0x1208,  # ለ
    0x1210,  # ሐ
    0x1218,  # መ
    0x1220,  # ሠ
    0x1228,  # ረ
    0x1230,  # ሰ
    0x1238,  # ሸ
    0x1240,  # ቀ
    0x1248,  # ቈ
    0x1250,  # ቐ
    0x1258,  # ቘ
    0x1260,  # በ
    0x1268,  # ቨ
    0x1270,  # ተ
    0x1278,  # ቸ
    0x1280,  # ኀ
    0x1288,  # ኈ
    0x1290,  # ነ
    0x1298,  # ኘ
    0x12A0,  # አ
    0x12A8,  # ከ
    0x12B0,  # ኰ
    0x12B8,  # ኸ
    0x12C0,  # ዀ
    0x12C8,  # ወ
    0x12D0,  # ዐ
    0x12D8,  # ዘ
    0x12E0,  # ዠ
    0x12E8,  # የ
    0x12F0,  # ደ
    0x12F8,  # ዸ
    0x1300,  # ጀ
    0x1308,  # ገ
    0x1310,  # ጐ
    0x1318,  # ጘ
    0x1320,  # ጠ
    0x1328,  # ጨ
    0x1330,  # ጰ
    0x1338,  # ጸ
    0x1340,  # ፀ
    0x1348,  # ፈ
    0x1350,  # ፐ
]

# Generate all characters (33 consonants × 7 vowel forms)
for base in base_consonants:
    for vowel_offset in range(7):  # 7 vowel forms
        char_code = base + vowel_offset
        amharic_chars.append(chr(char_code))

# Font path for rendering the characters
font_path = os.path.expanduser("~/.fonts/AbyssinicaSIL-Regular.ttf")

# Verify the font file exists
if not os.path.exists(font_path):
    print(f"Font file {font_path} does not exist! Please check the path and try again.")
    exit()

# List of background images in the 'background' subfolder
backgrounds = []
valid_extensions = ('.jpg', '.jpeg', '.png',)  # Valid image extensions
for bg_file in os.listdir('background'):
    if bg_file.lower().endswith(valid_extensions):
        backgrounds.append(os.path.join('background', bg_file))

# Verify that there are background images
if not backgrounds:
    print("No valid background images found in the 'background' subfolder! Please add images (e.g., .jpg, .png) and try again.")
    exit()

# Verify that the background images exist
for bg_path in backgrounds:
    if not os.path.exists(bg_path):
        print(f"Background image {bg_path} does not exist! Please add the background images to the 'background' subfolder and try again.")
        exit()

# Verify that the standard font folders exist
for char in amharic_chars:
    if not os.path.exists(char):
        print(f"Folder {char} does not exist! Please run the standard font generation script first.")
        exit()

# Generate images for each character with each background
for char in amharic_chars:
    try:
        font = ImageFont.truetype(font_path, size=40)
    except Exception as e:
        print(f"Font {font_path} not found or failed to load: {e}, exiting...")
        exit()

    for bg_idx, bg_path in enumerate(backgrounds):
        try:
            background = Image.open(bg_path).convert("L")  # Convert to grayscale
        except Exception as e:
            print(f"Background {bg_path} not found or failed to load: {e}, skipping...")
            continue

        # Generate 5 variations per background
        for i in range(5):
            # Resize background to 64x64
            bg_resized = background.resize((64, 64), Image.Resampling.LANCZOS)

            # Create a blank 64x64 image for the character
            char_image = Image.new("L", (64, 64), color=255)  # White background
            draw = ImageDraw.Draw(char_image)

            # Calculate text position with slight random offset
            text_bbox = draw.textbbox((0, 0), char, font=font)
            text_width = text_bbox[2] - text_bbox[0]
            text_height = text_bbox[3] - text_bbox[1]
            base_position = ((64 - text_width) // 2, (64 - text_height) // 2)
            offset_x = base_position[0] + random.randint(-3, 3)
            offset_y = base_position[1] + random.randint(-3, 3)
            position = (offset_x, offset_y)

            # Draw the character with a slight rotation
            temp_image = Image.new("L", (64, 64), color=255)
            temp_draw = ImageDraw.Draw(temp_image)
            temp_draw.text((32, 32), char, font=font, fill=0, anchor="mm")
            rotated_image = temp_image.rotate(random.randint(-5, 5))

            # Create a mask from the character image
            mask = rotated_image.point(lambda p: 255 if p == 0 else 0)

            # Overlay the character on the background
            final_image = Image.composite(rotated_image, bg_resized, mask)

            # Save the image in the same folder
            final_image.save(f"{char}/{char}_background_{bg_idx + 1}_{i + 1}.png")

print("Images with different backgrounds added to standard font folders successfully!")
```

```python
import os
import cv2
import numpy as np
from tensorflow.keras.preprocessing.image import ImageDataGenerator
from tensorflow.keras.utils import to_categorical
from tensorflow.keras.callbacks import EarlyStopping
import matplotlib.pyplot as plt
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Conv2D, MaxPooling2D, Flatten, Dense, Dropout
```

```python


class DataLoader:
    def __init__(self, data_dir, img_size=(64, 64)):
        self.data_dir = data_dir
        self.img_size = img_size
        self.classes = sorted(os.listdir(data_dir))
        self.num_classes = len(self.classes)
        self.class_to_idx = {cls: idx for idx, cls in enumerate(self.classes)}

    def load_data(self, target_num_classes=None):
        images = []
        labels = []
        
        for class_name in self.classes:
            class_dir = os.path.join(self.data_dir, class_name)
            for img_name in os.listdir(class_dir):
                img_path = os.path.join(class_dir, img_name)
                img = cv2.imread(img_path, cv2.IMREAD_GRAYSCALE)
                img = cv2.resize(img, self.img_size)
                images.append(img)
                labels.append(self.class_to_idx[class_name])
        
        images = np.array(images).reshape(-1, 64, 64, 1) / 255.0
        # Use target_num_classes if provided, otherwise use self.num_classes
        labels = to_categorical(labels, target_num_classes or self.num_classes)
        return images, labels

    def get_datagen(self):
        return ImageDataGenerator(
        rotation_range=20,  # Increased from 10
        width_shift_range=0.2,  # Increased from 0.1
        height_shift_range=0.2,  # Increased from 0.1
        zoom_range=0.2,  # Increased from 0.1
        shear_range=0.2,  # Add slight shearing
        fill_mode='nearest'
    )

```

```python



def build_model(num_classes):
    model = Sequential([
        Conv2D(32, (3, 3), activation='relu', input_shape=(64, 64, 1)),
      
        MaxPooling2D((2, 2)),
        Conv2D(64, (3, 3), activation='relu'),
       
        MaxPooling2D((2, 2)),
        Conv2D(128, (3, 3), activation='relu'),
       
        MaxPooling2D((2, 2)),
        Flatten(),
        Dense(256, activation='relu'),
        Dropout(0.5),  # Reduced from 0.5
        Dense(num_classes, activation='softmax')
    ])
    
    model.compile(optimizer='adam', loss='categorical_crossentropy', metrics=['accuracy'])
    return model
```

```python
def train_and_evaluate(model, datagen, X_train, y_train, X_val, y_val, X_test, y_test):
    history = model.fit(
        datagen.flow(X_train, y_train, batch_size=32),
        epochs=50,
        validation_data=(X_val, y_val),
        callbacks=[EarlyStopping(patience=10)]
    )
    test_loss, test_acc = model.evaluate(X_test, y_test)
    print(f"Test Accuracy: {test_acc:.4f}")
    
    plt.plot(history.history['accuracy'], label='Train Accuracy')
    plt.plot(history.history['val_accuracy'], label='Validation Accuracy')
    plt.title('Model Accuracy')
    plt.xlabel('Epoch')
    plt.ylabel('Accuracy')
    plt.legend()
    plt.show()
    return model
```

```python
from sklearn.model_selection import train_test_split

def main():
    # Load training data
    train_loader = DataLoader("dataset/T")
    X_train_full, y_train_full = train_loader.load_data()
    num_classes = train_loader.num_classes
    print(f"Training classes: {num_classes}")

    # Load test data with training's num_classes
    test_loader = DataLoader("dataset/te")
    X_test, y_test = test_loader.load_data(target_num_classes=num_classes)
    print(f"Test data shape: {X_test.shape}, y_test shape: {y_test.shape}")

    # Split train into train/validation
    X_train, X_val, y_train, y_val = train_test_split(
        X_train_full, y_train_full, test_size=0.2, random_state=42
    )
    
    # Build and train model
    model = build_model(num_classes)
    datagen = train_loader.get_datagen()
    model = train_and_evaluate(model, datagen, X_train, y_train, X_val, y_val, X_test, y_test)
    
    # Save model
   

    model.save("amharic_cnn.h5")
    print("Model saved as 'amharic_cnn.h5'")

if __name__ == "__main__":
    
    main()
```
